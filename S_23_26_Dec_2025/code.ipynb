# for-else Statement

The else block just after for/while is executed only when the loop is NOT terminated by a break statement.

```python
for i in range(1, 4):
    print(i)
else:  # Executed because no break in for
    print("No Break")
```

```python
for i in range(1, 4):
    print(i)
    break
else: # Not executed as there is a break
    print("No Break")
```

The for-else structure is highly useful for searching items in a collection because it removes the need for a temporary "flag" variable.

```python
# Look for an even number
numbers = [1, 3, 5, 7]

for num in numbers:
    if num % 2 == 0:
        print(f"Found an even number: {num}")
        break
else:
    print("No even numbers found in the list.")
```
