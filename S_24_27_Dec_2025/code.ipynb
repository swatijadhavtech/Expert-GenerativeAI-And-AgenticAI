# Functions

Python functions are reusable blocks of code designed to perform a specific, single task.


- Types
    - Inbuild Function
        - print(), len(), max(), head(), tail(), etc
    - User Defined Function

        - Syntax:

            def function_name(arguments):

                function body

                return value
- Argument Category
    - Formal Argument
    
        These are the variable names listed in the function definition.
    - Actual Argument

        These are the real values, expressions, or variables passed into the function when you execute it.
- Argument Types
    - Positional
    - Keyword
    - Default
    - Arbitrary ( *args )
    - Arbitrary Keyword ( **kwargs )

Function behavior depends on whether the object is mutable or immutable.

 - Mutable objects like lists can be modified inside functions.
 - Immutable objects like integers and strings remain unchanged.

Note: 

Python uses pass-by-object-reference, where functions receive references to objects instead of actual copies.

```python
def welcome():
    return f"Hello, Welcome to the program."

welcome()
```

```python
def greet(name, greeting):
    print(f"{greeting}, {name}!")

greet("Divya", "Good morning")
```

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 3)
print(f"The sum is: {result}")
```

```python
def add_sub(a, b):
    sum_result = a + b
    sub_result = a - b
    return sum_result, sub_result

result = add_sub(10, 5)
print(f"The sum is: {result[0]}")  
print(f"The difference is: {result[1]}")  
print(type(result))
```

Avoid using print() inside Python functions because it ruins their reusability by forcing output to the console instead of returning a usable value.

```python
def calculate_tax(price):
    tax = price * 0.1
    print(tax)  # Bad practice: Traps the value

# Try to use the result to find the final price
total_tax = calculate_tax(100)  # Prints 10.0
final_price = 100 + total_tax   # Crashes: TypeError (100 + None)
```

```python
def calculate_tax(price):
    tax = price * 0.1
    return tax

# Try to use the result to find the final price
total_tax = calculate_tax(100)  # Returns 10.0
final_price = 100 + total_tax   # Works: 110.0
print(f"The final price is: {final_price}")
```

```python
def person(name, age):
    return f"{name} is {age} years old."

print(person("Alice", 30))
```

# Keyword Argument

```python
def person(name, age):
    return f"{name} is {age} years old."

print(person(age=30, name="Alice"))
```

# Default Argument

```python
def person(name, age=20):
    return f"{name} is {age} years old."

print(person(name="Alice"))
```

# Arbitrary ( *args )

```python
def add_numbers(a, b):
    
    return a + b

result = add_numbers(5, 3 ,5, 7, 9)
print(f"The sum is: {result}")
```

```python
def add_numbers(a, *b):
    
    return a + b

result = add_numbers(5, 3 ,5, 7, 9)
print(f"The sum is: {result}")
```

```python
def add_numbers(a, *b):
    c = a
    for num in b:
        c += num
    return c

result = add_numbers(5, 3 ,5, 7, 9)
print(f"The sum is: {result}")
```

```python
def person(name, *data):
    print(name)
    print(data)
    
person("Alice", 30, "Engineer", "New York")
```

# Arbitrary Keyword ( **kwargs )

```python
def person(name, *data):
    print(name)
    print(data)
    
person("Alice", age=30, occupation="Engineer", city="New York")
```

```python
def person(name, **data):
    print(name)
    print(data)
    
person("Alice", age=30, occupation="Engineer", city="New York")
```

# Global Variable and Local Variable

```python
a = 10  # global variable

def something():
    b= 15
    print('in function',b)
    print('in function',a)
    
something()
```

```python
a = 10  # global variable

def something():
    b= 15
    print('in function',b)
    print('in function',a)
    
something()
print('out function',b)
print('out function',a)
```

```python
x = 10 # Global variable

def update_x():
    global x  # Declare that we are using the global variable x to modify it
    x += 5
    
update_x()
print(x)  # Output: 15
```
