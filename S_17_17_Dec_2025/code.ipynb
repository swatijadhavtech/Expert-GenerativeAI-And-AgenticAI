# PIL (Python Imaging Library)

PIL (Python Imaging Library) is the standard library used for opening, manipulating, and saving many different image file formats in Python.

Even though you use the code name PIL in your scripts, you must install the Pillow package


```python
pip install pillow
```

```python
from PIL import Image
```

r is for raw string to avoid issues with backslashes in file path

```python
horse_image = Image.open(r"C:\Users\SWATI\Desktop\horse.jpg") 
horse_image
```

```python
type(horse_image)
```

Look at its properties

```python
print(horse_image.format, horse_image.size, horse_image.mode)
```

Preview the image (opens in default system viewer)

```python
horse_image.show()
```
