# Number System

Number System is a system that defines numbers in different ways to represent numbers in different forms.

 - Binary Number System (base or radix =2) - 0b
 - Octal Number System (base or radix  = 8) - 0o
 - Decimal Number System (base or radix  = 10) - 0x
 - Hexadecimal Number System (base or radix  = 16) - 0xabcdef | a-10, b-11, c-12, d-13, e-14, f-15
