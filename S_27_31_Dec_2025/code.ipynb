# Univariate Analysis (EDA)

Univariate Analysis means analyzing one variable at a time to understand its distribution, pattern, and characteristics.

Objectives

 - Understand data distribution
 - Find central tendency (mean, median, mode)
 - Detect outliers
 - Identify missing values
 - Understand frequency of categories

Types of Univariate Analysis

  - Numerical Variables

  we use statistics like mean, median, and standard deviation along with histograms, box plots and density plot.

Example: Salary, Age, Marks

  - Categorical Variables

  we use frequency counts along with bar chart and pie chart.

Example: Gender, City, Department

# Bivariate Analysis (EDA)

Bivariate Analysis means analyzing two variables together to understand the relationship between them.

Objectives

- Find relationships between variables.
- Determine whether one variable affects another.
- Identify correlations, trends, or patterns.

Correlation

- Correlation is a statistical measure used in bivariate analysis to determine the strength and direction of the relationship between two numerical variables. 

- Its value ranges from -1 to +1, where +1 indicates a perfect positive relationship, -1 indicates a perfect negative relationship, and 0 indicates no relationship.

Correlation Coefficient (r)

The value of correlation ranges from -1 to +1.

Correlation Meaning And Value
- Perfect Positive Correlation (+1)
- Strong Positive Correlation (+0.7 to +0.9)
- Moderate Positive Correlation (+0.3 to +0.7)
- No Correlation (0)
- Moderate Negative Correlation (-0.3 to -0.7)
- Strong Negative Correlation (-0.7 to -0.9)
- Perfect Negative Correlation (-1)

# Outlier Treatment (EDA)

Outlier treatment is the process of identifying and handling extreme data points that differ significantly from the rest of the dataset.

Why Treat Outliers?

- Can skew the mean and standard deviation.
- May reduce model accuracy.
- Can affect regression models significantly.

Common detection methods include Box Plot, IQR, and Z-Score. 

Treatment techniques include removing outliers, capping them, applying transformations such as logarithms, or retaining them if they represent valid business cases.

# Missing Value Treatment (EDA)

Missing value treatment is the process of identifying and handling null or missing values during EDA.

First, Calculate the number and percentage of missing values and understand why they are missing. Then, depending on the data type and distribution, You may remove records, use mean or median for numerical variables, mode or an Unknown category for categorical variables, or use advanced imputation techniques such as KNN. You also make sure that the treatment doesn't introduce bias into the dataset.

# Variable Creation (EDA)

Variable creation means creating a new variable/feature from existing data to make the dataset more useful for analysis or machine learning.

It is also called Feature Engineering.

Why do we create variables?

Variable creation can:

- Improve model performance
- Extract useful information from existing columns
- Capture business relationships
- Convert raw data into meaningful features
- Make patterns easier to identify

# Variable Transformation (EDA)

Variable transformation means changing the scale, distribution, or representation of an existing variable without changing the underlying meaning of the data.

It is mainly done to:

- Reduce skewness
- Handle outliers
- Make variables comparable
- Improve model performance
- Satisfy assumptions of some statistical/ML algorithms
