# Variable Data Types 

Int - Int, or integer, is a whole number, positive or negative, without decimals, of unlimited length.

```python
i=5
type(i)
```

Float - Float, or "floating point number" is a number, positive or negative, containing one or more decimals. ( only e letter is allowed)

```python
f=11.45
type(f)
```

```python
f1=1e0
f1
```

```python
f2=1e1
f2
```

```python
f3=2.3e3
f3
```

bool - Booleans represent one of two values: True or False

```python
b=True
type(b) 
```

```python
b=True
b1=False
b+b1
```

```python
print(b+b1)
print(b-b1)
print(b*b1)
```

String - Strings are surrounded by either single or double quotation marks.

```python
s='prakash'
s
```

```python
type(s)
```

String Indexing - strings can be indexed using the same notation ([index]) as lists

```python
s[0]
```

```python
s[-1]
```

```python
s[10]
```

```python
s[-10]
```

String Slicing - a substring can be accessed using slicing ([start:end]).

```python
s[:] # print entire string
```

```python
s[2:] # print from index 2 to end
```

```python
s[:5] # print from start to index 5 (excluding 5)
```

```python
s[2:10] # print from index 2 to index 10 (excluding 10)
```

```python
s[0:6:2] # print from index 0 to 6 with step size 2
```

```python
s[-3]
```

```python
s[-2:]
```

```python
s[-7:-2]
```

```python
s[2:-2]
```
