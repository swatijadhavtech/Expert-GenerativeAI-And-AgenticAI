# Masking

Masking in NumPy is a technique used to filter, extract, modify, or manipulate specific elements in an array based on a conditional boolean test.

```python
mat = np.arange(0,100).reshape(10,10)
mat
```

```python
mat>50
```

```python
mat[mat>50]
```
