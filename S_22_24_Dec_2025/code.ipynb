# Conditional Statements

Conditional statements in Python allow you to execute different blocks of code based on whether a specific condition evaluates to True or False.
- Python relies on indentation (usually four spaces) rather than curly braces {} to define the scope of these code blocks.

# if Statement 

It runs a block of code only if its condition is true

```python
# number is even or odd
x =4
r = x % 2

if r == 0:
    print("Even Number")
```

```python
x =5
r = x % 2

if r == 0:
    print("Even Number")
```

```python
x =5
r = x % 2

if r == 0:
    print("Even Number")
if r != 0:
    print("Odd Number")
    
```

# else Statement

It executes only if the if condition evaluates to false

```python
x =4
r = x % 2

if r == 0:
    print("Even Number")
else:
    print("Odd Number") 
```

# Nested if Statement

```python
x =8
r = x % 2

if r == 0:
    print("Even Number")
    if x > 5:
        print("Greater than 5")
else:
    print("Odd Number") 
```

```python
x =3
r = x % 2

if r == 0:
    print("Even Number")
    if x > 5:
        print("Greater than 5")
    else:
        print("Less than 5")
else:
    print("Odd Number") 
```

# if-elif-else Ladder

Python stops checking remaining conditions as soon as it finds a True match.

```python
x = 10
if x == 1:
    print("One")
elif x == 2:
    print("Two")    
elif x == 3:
    print("Three")
elif x == 4:
    print("Four")
else:
    print("Greater than Four")
```

# Loops

Loops are to repeat a block of code multiple times, either by iterating over a sequence or while a condition remains true.

# while loops


Use a while loop to execute code as long as a conditional statement is true.

```python
i = 1
while i <= 5:
    print("data science")
    i += 1
```

# For Loops

Use a for loop to step through items in a sequence like lists, tuples, or strings.

```python
# Iterating over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

# Loop Control Statements
You can alter the flow of a loop using control keywords.

- break: Exits the loop immediately.
- continue: Skips the current iteration and moves to the next.
- pass: Acts as a placeholder doing absolutely nothing. 

```python
for num in range(1, 10):
    if num == 2:
        continue  # Skips 2
    if num == 6:
        break  # Stops the loop at 6
    print(num)  
```
