# Type Casting

Type Casting is the method to convert the Python variable datatype into a certain data type.

There can be two types of Type Casting in Python:

- Implicit Type Conversion (python automatically converts)


- Explicit Type Conversion (programmer manually converts)

complex(): Python complex() function take int or one string as an argument and return complex type object.

With one argument: The integer becomes the real part, and the imaginary part defaults to zero.

```python
complex(8)
```

With two arguments: The first argument is the real part, and the second argument is the imaginary part. Both can be integers (or floats).

```python
complex(8,10)
```

```python
complex(10,20,30)
```

```python
complex(2.3)
```

```python
complex(2.3,4.5)
```

```python
complex(True, False)
```

```python
complex(False, True)
```

```python
complex('10')
```

We can type cast from other datatypes to complex except 2 string parameter & text string

```python
complex('10','20')
```

```python
complex(10,'20')
```

```python
complex('10',20)
```

```python
complex('ten')
```

# print()

In Python, print formatting means displaying variables inside a string in a clean and readable way.

```python
a=10
b=20
a
b
```

```python
print(10)
print(10,20)
print('python')
print(10,20,'python')
```

```python
num1,num2=20,30
add=num1+num2
print(add)  
```

```python
num1,num2=20,30
add=num1+num2
print('The addition of', num1, 'and', num2, 'is:', add)
```

format() method - Placeholders {} are used in the string and replaced by values passed as arguments to the method.

```python
num1,num2=20,30
add=num1+num2
print('The addition of {0} and {1} is: {2}'.format(num1, num2, add))
```

f-Strings - f-strings provide a concise and readable way to embed expressions inside string literals.

```python
num1,num2=20,30
add=num1+num2
print(f'The addition of {num1} and {num2} is: {add}')
```

end parameter - By default, print() adds a newline but end lets you customize what appears after the output like a space, tab or nothing

```python
print('hello')
print('good morning')
```

```python
print('hello', end=' ')
print('good morning')
```

```python
print('*')
print('**',end='')
print('***')    
```

sep parameter - By default, print() adds a space but sep lets you customize what appears between the multiple values like a space, tab or nothing

```python
print('Hello','World',sep='---->')
```
