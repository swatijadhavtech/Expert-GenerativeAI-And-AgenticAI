# Numpy

NumPy (Numerical Python) is the foundational library for scientific computing in Python. It is primarily used to handle large, multi-dimensional arrays and matrices, offering high-performance tools that are significantly faster than standard Python lists. 

It has modules regarding various operations:
- Arithmetic operations
- Statistical Operations
- Bitwise Operators
- Linear Algebra
- Copying and viewing arrays
- Stacking
- Searching, Sorting, and counting, etc.
- Mathematical Operations
- Broadcasting
- Matplotlib for graphical representations
- Matrix Operations, etc.

The np.arange() function in NumPy creates an array of evenly spaced values within a given interval.

 Syntax: np.arange([start, ]stop, [step, ]dtype=None)

```python
np.arange(10)
```
