The shape property in pandas returns a tuple representing the dimensionality of a DataFrame or Series.

Behavior by Object Type
 - For a DataFrame: It returns a tuple in the format (rows, columns).
 - For a Series: It returns a tuple with a single value (rows,) because a Series is one-dimensional.

isnull() and isna() are same. Both functions return True for missing data types and False for valid data.
 - Detected as True: None, numpy.NaN (or np.nan), and pd.NA. 
 - Detected as False: Empty strings (""), white spaces (" "), and infinity values (numpy.inf).

```python
df.isnull().sum()
```
