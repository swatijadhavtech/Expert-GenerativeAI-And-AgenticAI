Formatted Output

Using format method

```python
num1=20
num2=30
sum=num1+num2   
print('The addition of {} and {} is={}'.format(num1, num2, sum) )
```

```python
name='Python'
age=20
city='New York'
print('Hello, my name is {} I am {} years old and I live in {}'.format(name, age, city))
```

```python
num1=100
num2=25
num3=333
avg=(num1+num2+num3)/3 
avg1=round((num1+num2+num3)/3,2)
print('The average of {}, {}, and {} is={} or {}'.format(num1, num2, num3, avg, avg1) )
```

 
 Using f-strings

```python
num1=20
num2=30
sum=num1+num2
print(f'The addition of {num1} and {num2} is={sum}' )
```

```python
name='Python'
age=20
city='New York'
print(f'Hello, my name is {name} I am {age} years old and I live in {city}')
```

```python
num1=100
num2=25
num3=333
avg=round((num1+num2+num3)/3,2)
print(f'The average of {num1}, {num2}, and {num3} is={avg}')
```

```python
num1=10
num2=20 
add=num1+num2
print('The addition of', num1, 'and', num2, 'is=',add)
print('The addition of {} and {} is={}'.format(num1, num2, add) )
print(f'The addition of {num1} and {num2} is={add}')
```

end - Specifies what to print at the end of the output. The default is a newline character ('\n')

```python
print('hello')
print('good morning')
```

```python
print('hello', end=' ')
print('world good day')
```

sep - Specifies the string to insert between multiple objects. The default is a single space (' ')

```python
print('hello','hai', 'how are you', sep='--->')
```

```python
print('hello','hai', 'how are you', sep='&')
```

```python
print('hello','hai', 'how are you', sep='@')
```

```python
print('hello','hai', 'how are you', sep=' ')
```

```python
print(3,'.') 
```

```python
print(3,'.',sep='') # in above example . is far from 3 so we will use sep method to remove space
```

```python
print(1,2,end=' ')
print(3,'.',sep='')
```
