# Using a Backslash for Explicit Line Continuation


You can use a backslash ( \ ) at the end of a physical line to tell Python that the line continues on the next one. This prevents a newline character from being added to the string at that point. 

```python
print("""\
    Usage: thingy [OPTIONS]
     -h                        Display this usage message
     -H hostname               Hostname to connect to
""")  # use triple quotes for multi-line strings
```

Indexing - It is used to obtain individual characters

```python
word = 'Python'
word[0]  # character in position 0
```

```python
word[5]
```

Note: Since -0 is the same as 0, negative indices start from -1

```python
word[-1]  # last character
```

```python
word[-2]  # second-last character
```

```python
word[-6]
```

Slicing - It is used to obtain a substring

```python
word[0:2]  # characters from position 0 (included) to 2 (excluded)
```

```python
word[2:5]  # characters from position 2 (included) to 5 (excluded)
```

An omitted frist index defaults to 0

```python
word[:2]  # characters from the beginning to position 2 (excluded)
```

An omitted second index defaults to the size of the string being sliced

```python
word[4:]  # characters from position 4 (included) to the end
```

```python
word[-2:]  # characters from the second-last to the end
```

Note: The start is always included and the end always excluded.

```python
word[:2] + word[2:]
```

```python
word[:4] + word[4:]
```

```python
word[42]  # the word only has 6 characters
```

```python
word[4:42]  # out of range slice indexes are handled gracefully
```

```python
word[42:] 
```

String cannot be changed --- they are immutable

```python
word[0] = 'j'
```
