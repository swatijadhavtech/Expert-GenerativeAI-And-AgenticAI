# Exploratory Data Analysis.

It is the process of analyzing and visualizing data to understand its structure, patterns, relationships, and anomalies before building models or drawing conclusions.

Steps of EDA

- Understanding the Business: Before analyzing data, you need to understand why the data exists and what business problem needs to be solved.

- Variable Identification: It is the process of understanding all columns (features) in the dataset and classifying them based on their type and role.
    
    - Independent Variable (Feature (X)):

        - The factor that influences or predicts an outcome.
        - It does not depend on another variable.

    - Dependent Variable (Target (Y)):

        - The outcome you want to measure or predict.
        - Its value depends on the independent variables.
        - It can be binary (Yes/No), categorical (multiple classes), or numerical (continuous values) depending on the business problem and type of analysis being performed.

            - Relevant Variables - Variables that have a meaningful relationship with the target variable and help improve predictions.
            - Irrelevant Variables - Variables that do not contribute to predicting the target variable or add little value.
