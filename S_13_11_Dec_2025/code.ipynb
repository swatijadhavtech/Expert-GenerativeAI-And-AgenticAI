# Dictionary

Dictionary is a data structure that stores information in key-value pairs.

An ordered collection of items that is mutable (changeable) with unique and immutable (unchangeable) keys.

You can create a dictionary using curly braces {} or the built-in dict() constructor. A value in a dictionary is accessed by using its key. This can be done either with square brackets [ ] or with the get() method.

Note: Accessing a missing key with [ ] raises a KeyError, while get() is safer because it returns None (or a default value) instead of an error.

Curly braces {} can be used to create a set or dict, but by default, they are treated as dictionaries. To declare an empty set, use the set() constructor.


```python
d = {}
type(d)
```

```python
s = set()
type(s)
```

```python
keys = {'a','b','c', 'd'}
mydict3 = dict.fromkeys(keys)
mydict3
```

```python
keys = {'a','b','c', 'd'}
value = 10
mydict3 = dict.fromkeys(keys, value)
mydict3
```

# Range

The range() function is a built-in tool used to generate a sequence of integers within a specified range.

- It only works with integers. Passing float values will result in a TypeError.
- You can access specific elements of a range by index (e.g., range(10)[2]).
- Check if a number is in a range using the in operator (e.g., 5 in range(10)).

range(start, stop, step)

- start (optional): The starting integer of the sequence. Defaults to 0.
- stop (required): The number at which the sequence stops. This value is exclusive, meaning the sequence goes up to but does not include this number.
- step (optional): The increment (or decrement) between each number. Defaults to 1.

```python
range(10)
```

```python
range(10,20)
```

```python
range(10,50,5)
```

```python
range(10,50,5,1)
```

If you need to see the full sequence or store it as a collection, you must explicitly convert the range object using the list() constructor.

```python
list(range(10))
```

```python
list(range(10,20))
```

```python
list(range(10,50,5))
```

```python
r = list(range(10,20))
for i in r:
    print(i)
```
