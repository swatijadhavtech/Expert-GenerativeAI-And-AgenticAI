# Data Cleaning Or Data Cleansing

```python
emp['Name']
```

Uppercase W replaces all non-word characters with an empty string

```python
emp['Name'] = emp['Name'].str.replace(r'\W', '', regex=True)  
emp
```

```python
emp['Domain'] = emp['Domain'].str.replace(r'\W', '', regex=True)  
emp
```

```python
emp['Location'] = emp['Location'].str.replace(r'\W', '', regex=True)  
emp
```

```python
emp['Salary'] = emp['Salary'].str.replace(r'\W', '', regex=True)  
emp
```

```python
emp['Age'] = emp['Age'].str.replace(r'\W', '', regex=True)  
emp
```

```python
emp['Age'] = emp['Age'].str.extract(r'(\d+)')
emp
```

```python
emp['Exp'] = emp['Exp'].str.extract(r'(\d+)')
emp
```

# 1 Missing Value Treatment

```python
clean_data['Age']
```

```python
import numpy as np
```

```python
clean_data['Age'] = clean_data['Age'].fillna(np.mean(pd.to_numeric(clean_data['Age'])))
clean_data
```

```python
clean_data['Exp'] = clean_data['Exp'].fillna(np.mean(pd.to_numeric(clean_data['Exp'])))
clean_data
```

```python
clean_data['Location'] = clean_data['Location'].fillna(np.mode(pd.to_numeric(clean_data['Location'])))
clean_data
```

```python
clean_data['Location'] = clean_data['Location'].fillna(clean_data['Location'].mode()[0])
clean_data
```
