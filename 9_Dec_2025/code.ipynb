# Tuple

An ordered collection of items that is immutable (Unchangeable), Heterogeneous (Different data types within a single structure) and allows duplicates members.

They are created by placing items inside parentheses () or without parentheses () and separated by commas. The comma itself is enough to define a tuple.

```python
t = ()
```

```python
t1 = (10,20,30,40,50,60)
t1
```

```python
type(t1)
```

```python
single_item = ("apple",)
# Without the comma, it's just a string: not_a_tuple = ("apple")
type(single_item)
```

You can convert other iterables (like lists or strings) into a tuple using the tuple() function.

```python
tuple_from_list = tuple(["apple", "banana", "cherry"])
type(tuple_from_list)
```

Elements in a tuple are accessed using indexing and slicing

```python
t1[0]
```

```python
t1[2:5]
```

Due to their immutability, tuples have fewer methods than lists, with the main ones being count() (to count occurrences) and index() (to find the position of a value)

```python
t1.count(20)
```

```python
t2= (10,20,30,10,20,10)
t2.count(10)
```

```python
t1.index(20)
```

```python
for i in t1:
    print(i)    
```

```python
for i in enumerate(t1):
    print(i)    
```
