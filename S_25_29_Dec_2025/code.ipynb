# Ollama

Ollama is a free, open-source framework that lets you download and run large language models (LLMs) directly on your own computer.

How Ollama Works
 - Local Processing: It runs as a lightweight service on your machine, using your CPU or GPU to generate AI responses locally.
 - Command-Line Control: You use simple text commands in your terminal to install, update, and run different models.
 - Built-in API: It provides a local REST API that lets software apps and code editors connect to your offline AI just like they would connect to cloud services.


Key Benefits
 - Complete Privacy: Your data and text prompts never leave your personal computer.
 - No Subscription Fees: It is entirely free to use after you download the software and models.
 - Works Offline: You do not need an active internet connection to chat or process data.

Download and install for window 

https://ollama.com/download/windows

Open cmd and run below command

 - ollama
 - ollama ls 

Install and run gemma3 model
 - ollama pull gemma3:270m
 - ollama run gemma3:270m

Then ask any question


Install and run deepseek model
 - ollama pull deepseek-r1:1.5b
 - ollama run deepseek-r1:1.5b

Then ask any question

Use ctrl + d or ctrl + z or /bye to exit
