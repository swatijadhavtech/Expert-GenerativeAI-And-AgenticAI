You can define a set using curly braces {} or the built-in set() constructor.

# Set Operation

```python
s3 = {1,2,3,4,5}
s4 = {4,5,6,7,8}
s5 = {8,9,10}
```

Union - All unique elements from both sets.

```python
s3.union(s4)
```

```python
s3 | s4 | s5
```

```python
s5.union(s3,s4)
```

Intersection - Elements present in both sets.

```python
s3.intersection(s4)
```

```python
s4 & s5
```

```python
s3 & s4 & s5
```

```python
s3.intersection(s4,s5)
```

Difference - Elements in s3 but not in s4.

```python
s3.difference(s4)
```

```python
s4 - s3
```

```python
s3 - s5
```

Symmetric Difference - Elements in either set, but not both.

It return a new set and do not modify the original sets.

```python
s3.symmetric_difference(s4)
```

```python
s4 ^ s5
```

```python
len(s3)
```

symmetric_difference_update - It modifies a set in-place by keeping only the elements found in either the original set or a specified iterable, but not in both.

It updates the original set directly rather than returning a new one.

```python
s3.symmetric_difference_update(s4)
s3
```

```python
len(s3)
```

```python
s3.difference_update(s4)
s3
```
