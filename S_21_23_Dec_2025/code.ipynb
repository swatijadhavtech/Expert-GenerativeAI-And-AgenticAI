# Data Visualisation

Data Visualisation is a graphical representation of information and data.
- Univariate Analysis: It is a type of data visualization where we visualize only a single variable at a time.
- Bivariate analysis: It is the simultaneous analysis of two variables.
- Multivariate Analysis: It is an extension of bivariate analysis which means it involves multiple variables at the same time to find correlation between them.

Need for Visualizing Data

- Understand the trends and patterns of data
- Analyze the frequency and other such characteristics of data
- Know the distribution of the variables in the data.
- Visualize the relationship that may exist between different variables

```python
import pandas as pd 
```

```python
df = pd.read_excel(r'F:\GenAI&AgeAI\Notes\Dec\Dec19 - Pandas\data.xlsx')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
plt.rcParams['figure.figsize'] = 6,3

import warnings
warnings.filterwarnings('ignore')
```

```python
vis =sns.displot(df['BirthRate'])
plt.show()
```

```python
vis =sns.displot(df['BirthRate'], bins=20, kde=True)
plt.show()
```

```python
vis =sns.boxplot(data=df, x='IncomeGroup', y='BirthRate')
plt.show()
```

Outlier:

An outlier is a data point that deviates (different) significantly from the rest of your dataset. 

```python
vis =sns.lmplot(data=df, x='InternetUsers', y='BirthRate')
plt.show()
```

```python
vis =sns.lmplot(data=df, x='InternetUsers', y='BirthRate', fit_reg=False, hue='IncomeGroup', legend=True)
plt.show()
```

```python
vis =sns.lmplot(data=df, x='InternetUsers', y='BirthRate', fit_reg=True, hue='IncomeGroup', legend=True)
plt.show()
```
